```python
conteudo_vendas = """101,Produto 1,3,1200.00,2024-01-10
102,Serviço 1,5,800.00,2024-01-11
103,Produto 2,2,2500.00,2024-01-12
104,Produto 3,4,600.00,2024-01-13
105,Serviço 2,10,150.00,2024-01-14
"""

with open("vendas.txt", "w", encoding="utf-8") as f:
    f.write(conteudo_vendas)

print("✅ vendas.txt criado com sucesso!")
```

```python
import json
import csv
import os
from datetime import datetime
from xml.etree.ElementTree import Element, SubElement, ElementTree, indent

# Configurações
ARQUIVO_ENTRADA = "vendas.txt"
ARQUIVO_JSON    = "vendas.json"
ARQUIVO_XML     = "vendas.xml"
DATA_IMPORTACAO = datetime.today().strftime("%d/%m/%Y")


def formatar_data(data_str):
    """Converte YYYY-MM-DD → DD/MM/YYYY (padrão brasileiro)."""
    try:
        dt = datetime.strptime(data_str.strip(), "%Y-%m-%d")
        return dt.strftime("%d/%m/%Y")
    except ValueError:
        return data_str.strip()


def ler_vendas(caminho):
    """Lê o arquivo TXT e retorna lista de dicionários com os dados tratados."""
    vendas = []
    if not os.path.exists(caminho):
        raise FileNotFoundError(f"Arquivo não encontrado: {caminho}")

    with open(caminho, newline="", encoding="utf-8") as f:
        leitor = csv.reader(f)
        for numero_linha, linha in enumerate(leitor, start=1):
            if len(linha) != 5:
                print(f"  ⚠️ Linha {numero_linha} ignorada — esperava 5 campos, encontrou {len(linha)}.")
                continue
            id_produto, nome_produto, quantidade, preco_unitario, data_venda = linha
            vendas.append({
                "id_produto"     : id_produto.strip(),
                "nome_produto"   : nome_produto.strip(),
                "quantidade"     : int(quantidade.strip()),
                "preco_unitario" : float(preco_unitario.strip()),
                "data_venda"     : formatar_data(data_venda),
                "data_importacao": DATA_IMPORTACAO,
            })
    return vendas


def exportar_json(vendas, caminho):
    """Exporta a lista de vendas para JSON formatado."""
    with open(caminho, "w", encoding="utf-8") as f:
        json.dump(vendas, f, ensure_ascii=False, indent=2)
    print(f"  ✅ JSON gerado: {caminho}")


def exportar_xml(vendas, caminho):
    """Exporta a lista de vendas para XML com indentação."""
    raiz = Element("vendas")
    for v in vendas:
        venda_elem = SubElement(raiz, "venda")
        SubElement(venda_elem, "id_produto"     ).text = v["id_produto"]
        SubElement(venda_elem, "nome_produto"   ).text = v["nome_produto"]
        SubElement(venda_elem, "quantidade"     ).text = str(v["quantidade"])
        SubElement(venda_elem, "preco_unitario" ).text = f"{v['preco_unitario']:.2f}"
        SubElement(venda_elem, "data_venda"     ).text = v["data_venda"]
        SubElement(venda_elem, "data_importacao").text = v["data_importacao"]

    indent(raiz, space="  ")
    arvore = ElementTree(raiz)
    with open(caminho, "wb") as f:
        arvore.write(f, encoding="utf-8", xml_declaration=True)
    print(f"  ✅ XML gerado: {caminho}")
    # --- Execução ---
print("=" * 55)
print("  LabDist Científica Ltda. — Integração de Vendas")
print(f"  Data de importação: {DATA_IMPORTACAO}")
print("=" * 55)

print(f"\n[1/3] Lendo: {ARQUIVO_ENTRADA}")
vendas = ler_vendas(ARQUIVO_ENTRADA)
print(f"      {len(vendas)} venda(s) carregada(s).")

print("\n[2/3] Exportando para JSON...")
exportar_json(vendas, ARQUIVO_JSON)

print("\n[3/3] Exportando para XML...")
exportar_xml(vendas, ARQUIVO_XML)

print("\n🎉 Integração concluída com sucesso!")
print("=" * 55)

# Exibe prévia do JSON gerado
print("\n📄 Prévia do vendas.json:")
with open(ARQUIVO_JSON, encoding="utf-8") as f:
    print(f.read())
```
