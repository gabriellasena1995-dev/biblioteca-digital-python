```python
import os

estrutura = [
    "documentos/pdf/2023",
    "documentos/pdf/2024",
    "documentos/epub/2023",
    "documentos/epub/2024",
    "documentos/docx/2023",
    "documentos/docx/2024"
]

for pasta in estrutura:
    os.makedirs(pasta, exist_ok=True)

print("Estrutura criada com sucesso!")
```

```python
import os

def listar_documentos():
    print("\n=== DOCUMENTOS CADASTRADOS ===")

    for raiz, diretorios, arquivos in os.walk("documentos"):

        if arquivos:
            print(f"\nPasta: {raiz}")

            for arquivo in arquivos:
                print(f" - {arquivo}")


def adicionar_documento(nome, tipo, ano):

    caminho = f"documentos/{tipo}/{ano}/{nome}"

    with open(caminho, "w", encoding="utf-8") as arquivo:
        arquivo.write("Documento criado para teste")

    print(f"Documento {nome} criado com sucesso.")


def renomear_documento(tipo, ano, nome_antigo, nome_novo):

    antigo = f"documentos/{tipo}/{ano}/{nome_antigo}"
    novo = f"documentos/{tipo}/{ano}/{nome_novo}"

    os.rename(antigo, novo)

    print("Documento renomeado com sucesso.")


def remover_documento(tipo, ano, nome):

    caminho = f"documentos/{tipo}/{ano}/{nome}"

    os.remove(caminho)

    print("Documento removido com sucesso.")
```

```python
adicionar_documento(
    "livro_python.pdf",
    "pdf",
    "2024"
)
```

```python
listar_documentos()
```

```python
renomear_documento(
    "pdf",
    "2024",
    "livro_python.pdf",
    "python_basico.pdf"
)
```

```python
listar_documentos()
```

```python
remover_documento(
    "pdf",
    "2024",
    "python_basico.pdf"
)
```

```python
listar_documentos()
```
